```python
import openai

client = openai.OpenAI()
memories = []


def setMessage(message):
    memories.append({"role": "user", "content": message})


def getAnswer():
    response = client.chat.completions.create(model="gpt-4o-mini", messages=memories)

    answer = response.choices[0].message.content
    print(f"AI: {answer}")

    memories.append({"role": "assistant", "content": answer})
```

```python
while True:
    message = input("입력하세요!")
    print(f"ME: {message}")

    if message == "quit" or message == "q":
        break
    else:
        setMessage(message)
        getAnswer()
```
