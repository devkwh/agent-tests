```python
import openai

PROMPT_FUNCTIONS = """
BASE_URL = "https://nomad-movies-2.nomadcoders.workers.dev"

/movies
인기 영화 목록
/movies/:id
ID로 영화 상세 정보 조회
/movies/:id/credits
영화의 출연진 및 제작진 조회
/movies/:id/similar
유사한 영화 조회

get_popular_movies() - /movies에서 인기 영화를 가져옵니다.
get_movie_details(id) - /movies/:id에서 영화 정보를 가져옵니다.
get_movie_credits(id) - /movies/:id/credits에서 출연진 및 제작진을 가져옵니다.

시스템은 3개의 함수를 사용할 수 있습니다.
질문에 대해서 어떤 함수를 어떻게 호출할지 답변해야 합니다.
질문은 여러개 일 수 있으며, 각 질문에 답변해야 합니다.
"""

PROMPT = """
질문: 지금 인기 있는 영화가 무엇인지 알려줘
질문: movie ID 550에 해당하는 영화가 무엇인지 알려줘
질문: movie ID 550에 해당하는 영화에 누가 출연하는지 알려줘
"""

client = openai.OpenAI()

response = client.chat.completions.create(
    model = "gpt-4o-mini",
    messages = [{"role": "user", "content": PROMPT_FUNCTIONS}, {"role": "user", "content": PROMPT}]
)

message = response.choices[0].message.content
message
```
